```python
import pandas as pd
import os 
os.getcwd()
```

This script will iterate through all CSV files in the specified folder "./DATA/Raw_Data/". 
For each file, it reads the data into a pandas DataFrame, prints the list of columns, and identifies potential primary key columns by checking which columns contain only unique and non-null values.

If we have a dataset whithout a pk output, it is necessary to create one to ensure data integrity and enable reliable joins with other tables.
If we have more than 1 in the output, it can be a composite key or a pk with a fk.


```python
path = "./DATA/Raw_Data/"

for file_name in os.listdir(path):
    if file_name.endswith(".csv"):
        file_path = os.path.join(path, file_name)
        df = pd.read_csv(file_path)
        print(f"\nColumns in {file_name}:")
        print(list(df.columns))
        ppk = []
        for col in df.columns:
            if df[col].is_unique and df[col].notnull().all():
                ppk.append(col)
        print(f"Primary key(s) in {file_name}: {ppk}")
```

```python
df = pd.read_csv("./DATA/Raw_Data/olist_orders_dataset.csv")
customer_order_counts = df.groupby('customer_id')['order_id'].count().reset_index()
customer_order_counts.rename(columns={'order_id': 'total_orders'}, inplace=True)
print(customer_order_counts.head())
```

```python
repeat_customers = customer_order_counts[customer_order_counts['total_orders'] > 1]
print(repeat_customers)
```
